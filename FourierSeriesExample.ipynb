# Fourier Series Example:

## Square Wave:

```python
import numpy as np
import matplotlib.pyplot as plt

# Time domain
x = np.linspace(-np.pi, np.pi, 1000)

# Initialize the Fourier series approximation
N_terms = 5  # Number of harmonics to include
y = np.zeros_like(x)

plt.figure(figsize=(8, 5))

# Add harmonics one by one
for n in range(1, 2 * N_terms, 2):  # Only odd harmonics
    y += (4 / (np.pi * n)) * np.sin(n * x)
    plt.plot(x, y, label=f'{n} harmonic{"s" if n > 1 else ""}')

plt.title(f'Fourier Series Approximation of Square Wave ({N_terms} terms)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

"Harmonic Analysis is the study of how a function f(x) can be decomposed into different type of basis "wave" functions. It is similar to the concept of a Taylor series and directly generalizes the concept of a trigonometric Fourier series. A Taylor series expresses f(x) as an infinite sum of polynomials {1, x, x2 ,...}. A trigonometric Fourier series expresses f(x) as an infinite sum of sines and cosines {sin(pix}, cos(pix}, sin(2pix}, cos(2pix}, sin(3pix}, cos(3pix},....}. Harmonic analysis generalizes the type of basis functions f(x) can be decomposed into. Another way to express f(x) is through a Fourier-Bessel series. This expresses f(x) as an infinite sum of Bessel functions. There are actually an infinite number of ways you can decompose f(x) into a sum of basis functions. These basis functions are usually obtained as the solutions of a class of differential equations called Sturm-Liovuille problems. Their solutions have certain orthogonality and eigenvalue properties that make harmonic decomposition possible. Sturm-Liouville problems and harmonic analysis are intimately related to the solutions of partial differential equations. Many of the laws of physics are formulated in terms of partial differential equations. It is for this reason much of 19th century mathematics was concerned with these topics."
